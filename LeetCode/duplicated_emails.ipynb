```python
import pandas as pd
```

```python
data = [[1, 'a@b.com'], [2, 'c@d.com'], [3, 'a@b.com']]
person = pd.DataFrame(data, columns=['id', 'email']).astype({'id':'Int64', 'email':'object'})
```

## Solutions

### Method 1

```python
def duplicate_emails1(person: pd.DataFrame) -> pd.DataFrame:
    # 1. Filter for rows that are duplicates
    # 2. Select only the 'email' column
    # 3. Drop duplicate rows so each email appears only once
    # 4. Convert the Series to a DataFrame
    df = person.loc[person.duplicated('email'), ['email']]
    
    # 5. Rename the column to match the expected capitalized 'Email'
    df = df.rename(columns={'email': 'Email'})
    
    return df

method1 = duplicate_emails1(person)
method1
```

### Method 2

```python
def duplicate_emails2(person: pd.DataFrame) -> pd.DataFrame:
    # 1. Count how many times each email appears
    counts = person['email'].value_counts()
    
    # 2. Filter for emails that appear more than once
    duplicate_emails = counts[counts > 1]
    
    # 3. Convert the index (which holds the email strings) into a DataFrame
    df = duplicate_emails.index.to_frame(index=False, name='Email')
    
    return df

method2 = duplicate_emails2(person)
method2
```

### Method 3

```python
def duplicate_emails3(person: pd.DataFrame) -> pd.DataFrame:
    df = person[person.duplicated(subset=['email'])] # find the duplicated rows in email and return the WHOLE ROW
    return df[['email']].drop_duplicates()
```
