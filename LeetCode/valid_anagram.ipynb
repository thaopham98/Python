```python
s="racecar"
t="carracv"
```

```python
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        freq = {}

        for c in s:
            freq[c] = freq.get(c, 0) + 1

        for d in t:
            if not d in freq.keys():
                return False

            freq[d] -= 1
            if freq[d] == 0:
                freq.pop(d)

        return len(freq) == 0
        
```

```python
solution1 = Solution()
solution1.isAnagram(s = "racecar" , t = "carrace")
```

```python
class Solution1:
    def isAnagram1(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        print(f"List s: {sorted(s)}")
        print(f"List t: {sorted(t)}")
        return sorted(s) == sorted(t)


    def isAnagram2(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False

        for i in set(s):
            if s.count(i) != t.count(i):
                return False

        return True
```
